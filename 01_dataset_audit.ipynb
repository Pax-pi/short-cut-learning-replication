```python
from pathlib import Path
import pandas as pd
```

```python
test_img_path = DATA_ROOT / metadata.img_filename[0]
```

```python
from matplotlib.image import imread
import matplotlib.pyplot as plt

test_img = imread(test_img_path)
```
